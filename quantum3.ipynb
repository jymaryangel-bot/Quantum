```python
from qiskit import QuantumCircuit
import numpy as np

# Create a circuit with 3 qubits
qc = QuantumCircuit(3)

# Single-qubit gates
qc.h(0)                     # Hadamard
qc.x(1)                     # Pauli-X
qc.y(2)                     # Pauli-Y
qc.z(0)                     # Pauli-Z

# Rotation gates
qc.rx(np.pi/4, 0)           # Rotate around X-axis
qc.ry(np.pi/3, 1)           # Rotate around Y-axis
qc.rz(np.pi/2, 2)           # Rotate around Z-axis

# Two-qubit gates
qc.cx(0, 1)                 # Controlled-X (CNOT)
qc.cz(1, 2)                 # Controlled-Z
qc.swap(0, 2)               # Swap qubits

# Add measurements
qc.measure_all()

# Draw the circuit
print(qc.draw("text"))
```
